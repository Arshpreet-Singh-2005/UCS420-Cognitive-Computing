## Q1: 1-D Array — Basic Operations\nCreate a NumPy 1D array with 5 elements and perform: (a) add 2 to all elements, (b) multiply all elements by 3, (c) divide every element by 2.

```
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)

# a) Addition of 2 in all the elements
added = arr + 2
print("After adding 2:", added)

# b) Multiply 3 with all the elements
multiplied = arr * 3
print("After multiplying by 3:", multiplied)

# c) Divide every element by 2
divided = arr / 2
print("After dividing by 2:", divided)
```

## Q2: Basic NumPy Array Questions

### a) Reverse the array

```
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("Original array:", arr)
print("Reversed array:", reversed_arr)
```

### b) Most frequent value and its indices

```
def most_frequent_with_indices(a):
    values, counts = np.unique(a, return_counts=True)
    max_count = counts.max()
    most_frequent_value = values[np.argmax(counts)]
    indices = np.where(a == most_frequent_value)[0]
    return most_frequent_value, indices, max_count


# i.
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
val_x, idx_x, count_x = most_frequent_with_indices(x)
print(f"x: most frequent value = {val_x} (occurs {count_x} times), indices = {idx_x}")

# ii.
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
val_y, idx_y, count_y = most_frequent_with_indices(y)
print(f"y: most frequent value = {val_y} (occurs {count_y} times), indices = {idx_y}")
```

## Q4: linspace, Array Properties, and Transpose

```
# 1-D array with 25 evenly spaced numbers from 10 to 100
Arshpreet_Singh = np.linspace(10, 100, 25)
print("Array:\n", Arshpreet_Singh)

print("\nDimensions (ndim):", Arshpreet_Singh.ndim)
print("Shape:", Arshpreet_Singh.shape)
print("Total elements (size):", Arshpreet_Singh.size)
print("Data type:", Arshpreet_Singh.dtype)
print("Total bytes consumed (nbytes):", Arshpreet_Singh.nbytes)
```

```
# Transpose using reshape()
# A 1-D array has only one axis, so a "true" transpose needs it reshaped
# into a 2-D array first (e.g. a row vector -> column vector).
transposed_via_reshape = Arshpreet_Singh.reshape(25, 1)
print("Transposed (via reshape to column vector), shape:", transposed_via_reshape.shape)
print(transposed_via_reshape[:5])
```

**Can we do the same with the `.T` attribute?**

Not directly on the original 1-D array — using `.T` on a 1-D array has **no effect**,
because `.T` reverses the order of axes, and a 1-D array only has one axis to begin
with, so there's nothing to reverse. This is demonstrated below.

However, if we first reshape the array into a 2-D array (e.g. shape `(25, 1)`), `.T`
on that 2-D version **does** work and produces the same result as `reshape()` did above.


```
# .T on the original 1-D array has no effect
print("Original shape:", Arshpreet_Singh.shape)
print("Shape after .T on 1-D array:", Arshpreet_Singh.T.shape)  # unchanged

# .T DOES work once the array is 2-D
reshaped_2d = Arshpreet_Singh.reshape(25, 1)
transposed_via_T = reshaped_2d.T
print("\nShape of 2-D reshaped array:", reshaped_2d.shape)
print("Shape after .T on the 2-D array:", transposed_via_T.shape)

print("\nSame result from reshape() and .T?",
      np.array_equal(transposed_via_reshape.reshape(1, 25), transposed_via_T))
```

## Q5: 2-D Array — Statistics, Reshape, and Resize

```
ucs420_arshpreet_singh = np.array([[10, 20, 30, 40],
                                    [50, 60, 70, 80],
                                    [90, 15, 20, 35]])
print("ucs420_arshpreet_singh:\n", ucs420_arshpreet_singh)
print("Shape:", ucs420_arshpreet_singh.shape)
```

```
mean_val = np.mean(ucs420_arshpreet_singh)
median_val = np.median(ucs420_arshpreet_singh)
max_val = np.max(ucs420_arshpreet_singh)
min_val = np.min(ucs420_arshpreet_singh)
unique_vals = np.unique(ucs420_arshpreet_singh)

print("Mean:", mean_val)
print("Median:", median_val)
print("Max:", max_val)
print("Min:", min_val)
print("Unique elements:", unique_vals)
```

```
# Reshape to 4 rows x 3 columns (reshape requires same total number of elements: 12 = 4*3)
reshaped_ucs420_arshpreet_singh = ucs420_arshpreet_singh.reshape(4, 3)
print("reshaped_ucs420_arshpreet_singh (4x3):\n", reshaped_ucs420_arshpreet_singh)
```

```
# Resize to 2 rows x 3 columns (2*3 = 6 elements)
# np.resize() allows changing to a smaller/larger total size (unlike reshape);
# here it keeps only the first 6 elements in row-major order.
resized_ucs420_arshpreet_singh = np.resize(ucs420_arshpreet_singh, (2, 3))
print("resized_ucs420_arshpreet_singh (2x3):\n", resized_ucs420_arshpreet_singh)
```
