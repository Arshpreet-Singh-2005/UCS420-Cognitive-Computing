## Question 1: Lists

```python
# Roll number digits, each multiplied by 10, to form list L
roll_number = "1024170170"
digits = [int(d) for d in roll_number]
L = [d * 10 for d in digits]
print("Digits of roll number:", digits)
print("L =", L)
```

## Question 2: Tuples

```python
scores = tuple(L[:8]) if len(L) >= 8 else tuple((L + L)[:8])
scores = tuple(digits[i] * 10 for i in range(8))
print("scores =", scores)
```

**(i) Highest score & its index; lowest score & how many times it appears**

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)

print(f"Highest score: {highest} at index {highest_index}")
print(f"Lowest score: {lowest}, appears {lowest_count} time(s)")
```

**(ii) Reverse the tuple (as a list)**

```python
first_score, second_score, *remaining_scores = scores
print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

## Question 3: Random Numbers

```python
import random
random.seed(1024170170)
random_numbers = [random.randint(100, 900) for _ in range(100)]
print(random_numbers)
```

**(ii) Count odd numbers**

```python
odd_numbers = [n for n in random_numbers if n % 2 != 0]
print("Count of odd numbers:", len(odd_numbers))
```

**(iii) Count even numbers**

```python
even_numbers = [n for n in random_numbers if n % 2 == 0]
print("Count of even numbers:", len(even_numbers))
```

**(iv) Count and list prime numbers**

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_numbers = [n for n in random_numbers if is_prime(n)]
print("Count of prime numbers:", len(prime_numbers))
print("Prime numbers found:", prime_numbers)
```

**(v) Most frequently occurring number**

```python
from collections import Counter
counts = Counter(random_numbers)
most_common_num, freq = counts.most_common(1)[0]
print(f"Most frequent number: {most_common_num}, occurs {freq} time(s)")
```
